## NOMAD Parser Tutorial 2 - Parsing using an ELN entry

This notebook tests the parser we build in the plugin tutorial 2. It includes the following steps:

1. Set up an `EntryArchive` section called `archive`. (Mimicks creation of an entry in the GUI)

2. Instantiate `OpticalMicroscopy` section, populate the `data_file` quantity with raw file path, and assign the section to `archive.data`.

3. Run the normalizer to parse the data from the raw file into the archive. (Mimicks saving an entry in the GUI)

4. Inspect the parsed archive: settings and results.

__NOTE:__ Before running this notebook, make sure only the entry points of Parser Tutorial 2 are uncommented in `pyproject.toml` and environment is reset.

### Step 1. Set up the entry archive

```python
import os

from nomad.datamodel.context import ClientContext
from nomad.datamodel.datamodel import EntryArchive

# Local directory mimicking the Upload/Project's raw directory
local_dir = os.path.abspath('../data/om_example')

# Let's build a dummy NOMAD archive that mimicks a real NOMAD entry archive that can be
# created by a user in the GUI.
archive = EntryArchive(
    metadata={
        'entry_name': 'optical_microscopy_test',
        'entry_type': 'OpticalMicroscopy',
    },
    m_context=ClientContext(local_dir=local_dir),
)
```

### Step 2. Populate `archive.data` with minimal data

```python
from nomad_plugin_tutorials.parsers.tutorial_2.schema.schema_package import (
    OpticalMicroscopy,
)

archive.data = OpticalMicroscopy(data_file='metadata.xml')
```

### Step 3. Run normalization

```python
from nomad.client import normalize_all

normalize_all(archive)
```

### Step 4. Inspect the parsed archive

```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt

print(f'Image resolution @ `archive.data.settings`: {archive.data.settings.resolution}')
print(f'Magnification @ `archive.data.settings`: {archive.data.settings.magnification}')

print('Image @ `archive.data.results[0]`:')
with archive.m_context.raw_file(archive.data.results[0].image) as file:
    img = mpimg.imread(file.name)
imgplot = plt.imshow(img)
```
