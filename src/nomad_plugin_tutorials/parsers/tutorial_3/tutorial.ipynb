## NOMAD Parser Tutorial 3 - Matching and creating an ELN entry (Hybrid Approach)

This notebook tests the parser we build in the plugin tutorial 3. It includes the following steps:

1. Parse the raw file using the `nomad.client.parse` function. (Mimicks uploading a file on the GUI). The file is matched with `ElnMatchingParser` and an archive for the raw file is generated. An `archive.json` file containing the input for the ELN archive (including the path of the raw data file) is also created.

2. Inspect the raw file archive: reference to ELN archive, metadata about raw file.

3. Parse the `archive.json` and run the normalizer to parse the data from the raw file into the archive. (Mimicks processing an entry when an `archive.json` file is dropped in the GUI)

4. Inspect the parsed ELN archive: settings and results.

__NOTE:__ Before running this notebook, make sure only the entry points of Parser Tutorial 3 are uncommented in `pyproject.toml` and environment is reset.

### Step 1. Parse the raw file into the raw file archive

```python
from nomad.client import parse

raw_file_archive = parse('../data/om_example/metadata.xml')[0]
```

### Step 2. Inspect the raw file archive

```python
print(
    f'ELN reference @ `raw_file_archive.data`: {raw_file_archive.data.eln}'
)
print(
    f'Data type @ `raw_file_archive.data`: {raw_file_archive.data.data_type}'
)
print(
    f'File ending @ `raw_file_archive.data`: {raw_file_archive.data.file_ending}'
)
print(
    f'File size @ `raw_file_archive.data`: {raw_file_archive.data.file_size}'
)
```

### Step 3. Parse ELN archive and run its normalization

```python
from nomad.client import normalize_all

eln_archive = parse('../data/om_example/metadata.archive.json')[0]
normalize_all(eln_archive)
```

### Step 4. Inspect the parsed ELN archive

```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt

print(
    f'Image resolution @ `archive.data.settings`: {eln_archive.data.settings.resolution}'
)
print(
    f'Magnification @ `archive.data.settings`: {eln_archive.data.settings.magnification}'
)

print('Image @ `archive.data.results[0]`:')
with eln_archive.m_context.raw_file(eln_archive.data.results[0].image) as file:
    img=mpimg.imread(file.name)
imgplot = plt.imshow(img)
```
