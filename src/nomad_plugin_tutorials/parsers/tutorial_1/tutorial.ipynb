## NOMAD Parser Tutorial 1 - Matching and creating a static entry

This notebook tests the parser we build in the plugin tutorial 1. It includes the following steps:

1. Set up an `EntryArchive` section called `archive`. (Mimicks creation of an entry in the GUI)

2. Parse the raw file using the `nomad.client.parse` function. (Mimicks uploading a file on the GUI). The file is matched with its parser and an archive is generated based on the parse logic.

3. Inspect the parsed archive: settings and results.

### Step 1. Set up the entry archive

```python
from nomad.datamodel.datamodel import EntryArchive

# Let's build a dummy NOMAD archive that mimicks a real NOMAD entry archive that can be
# created by a user in the GUI.
archive = EntryArchive(
    metadata={
        'entry_name': 'blackbody_test',
        'entry_type': 'BlackbodyRadiation',
    },
)
```

### Step 2. Parse the raw file into an entry archive

```python
from nomad.client import parse

archive = parse('../data/om_example/metadata.xml')[0]
```

### Step 3. Inspect the parsed archive

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

print(
    f'Image resolution @ `archive.data.settings`: {archive.data.settings.resolution}'
)
print(
    f'Magnification @ `archive.data.settings`: {archive.data.settings.magnification}'
)

print('Image @ `archive.data.results[0]`:')
img=mpimg.imread(archive.data.results[0].image)
imgplot = plt.imshow(img)
```
