## NOMAD Schema Tutorial - Planck's Blackbody Radiation

This notebook tests the schema we build in the schema package tutorial. It includes the following steps:

1. Set up an `EntryArchive` section called `archive`. (Mimicks creation of an entry in the GUI)

2. Instantiate the `BlackbodyRadiation` section with minimal user input and assign it to the `archive.data` section.

3. Run the normalizer to compute derived data from user inputs. (Mimicks saving an entry in the GUI)

4. Inspect the normalized archive: scalar results, arrays, and `archive.results`

5. Visualise the stored Plotly figure

### Background

The example schema (`BlackbodyRadiation`) takes a single temperature value as
input and computes the full spectral radiance profile B(λ, T) using Planck's
law:

$$B(\lambda, T) = \frac{2hc^2}{\lambda^5} \cdot \frac{1}{e^{hc / \lambda k_B T} - 1}$$

### Step 1. Set up the entry archive

```python
from nomad.datamodel.datamodel import EntryArchive

# Let's build a dummy NOMAD archive that mimicks a real NOMAD entry archive that can be
# created by a user in the GUI.
archive = EntryArchive(
    metadata={
        'entry_name': 'blackbody_test',
        'entry_type': 'BlackbodyRadiation',
    },
)
```

### Step 2. Populate the `archive.data` with minimal data

```python
from nomad_plugin_tutorials.schema.schema_package import BlackbodyRadiation

archive.data = BlackbodyRadiation(
    name='Molten Iron',
    temperature=1800.0,
)
```

### Step 3. Run normalization

```python
from nomad.client import normalize_all

normalize_all(archive)
```

### Step 4. Inspect the normalized archive

```python
print(
    f'Peak Wavelength @ `archive.data.results`: {archive.data.results.peak_wavelength}'
)
print(
    'Material Name @ `archive.results.material`: '
    f'{archive.results.material.material_name}'
)
```

### Step 5. Visualize the generated Plotly figure

```python
import plotly.graph_objects as go

print(
    'Custom Plotly figure added by the data normalizer'
    ' @ `archive.data.results.figures[0]`:'
)
fig = go.Figure(archive.data.results.figures[0].figure)
fig.update_layout(width=600, height=400)
fig.show()
```
